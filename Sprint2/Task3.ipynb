```polyglot-notebook
using System;
using System.Collections.Generic;

// Создание базового класса Employee

public class Employee
{
    // Поля класса
    private string name;
    private string position;
    private decimal salary;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string Position
    {
        get { return position; }
        set { position = value; }
    }

    public decimal Salary
    {
        get { return salary; }
        set
        {
            if (value >= 0)
                salary = value;
        }
    }

    // Конструктор класса
    public Employee(string name, string position, decimal salary)
    {
        this.name = name;
        this.position = position;
        this.salary = salary;
    }

    // Метод для вывода информации
    public virtual void PrintInfo()
    {
        Console.WriteLine($"Имя: {name}");
        Console.WriteLine($"Должность: {position}");
        Console.WriteLine($"Зарплата: {salary}");
    }
}

// Простое наследование

public class Manager : Employee
{
    // Дополнительное поле для отдела
    private string department;

    // Свойство для доступа к полю department
    public string Department
    {
        get { return department; }
        set { department = value; }
    }

    // Конструктор, который вызывает конструктор базового класса
    public Manager(string name, string position, decimal salary, string department)
        : base(name, position, salary) // Вызов конструктора Employee
    {
        this.department = department;
    }

    // Переопределение метода для вывода информации
    public override void PrintInfo()
    {
        base.PrintInfo(); // Сначала вызываем метод базового класса
        Console.WriteLine($"Отдел: {department}"); // Затем добавляем новую информацию
    }
}

// Интерфейс для создания отчетов
public interface IReportable
{
    string GenerateReport();
}

// Интерфейс для лидера команды
public interface ITeamLeader
{
    List<string> GetTeamMembers();
}

// Сложное наследование и реализация интерфейсов

public class ProjectManager : Manager, IReportable, ITeamLeader
{

    private string projectName;
    private List<string> teamMembers; // Поле для хранения членов команды

    public string ProjectName
    {
        get { return projectName; }
        set { projectName = value; }
    }

    // Конструктор, который вызывает конструктор базового класса Manager
    public ProjectManager(string name, string position, decimal salary, string department, string projectName)
        : base(name, position, salary, department) // Вызов конструктора Manager
    {
        this.projectName = projectName;
        this.teamMembers = new List<string>(); // Инициализируем список команды
    }

    // Переопределение метода для вывода информации
    public override void PrintInfo()
    {
        base.PrintInfo(); // Вызываем метод класса Manager
        Console.WriteLine($"Проект: {projectName}"); // Добавляем новую информацию
    }

    // Вспомогательный метод для добавления члена команды
    public void AddTeamMember(string memberName)
    {
        teamMembers.Add(memberName);
        Console.WriteLine($"В команду проекта «{projectName}» добавлен {memberName}.");
    }

    // Реализация метода из интерфейса IReportable
    public string GenerateReport()
    {
        return $"Отчет по проекту «{projectName}».\nМенеджер: {Name}.\nСостояние: В процессе.";
    }

    // Реализация метода из интерфейса ITeamLeader
    public List<string> GetTeamMembers()
    {
        return teamMembers;
    }
}

// Тестирование

// --- СОЗДАНИЕ ЭКЗЕМПЛЯРОВ КЛАССОВ ---

Employee сотрудникИван = new Employee(
    "Иван",
    "Разработчик",
    80000
);

Manager менеджерЕгор = new Manager(
    "Егор",
    "Начальник отдела",
    120000,
    "Отдел веб-разработки"
);

ProjectManager проектныйМенеджерАртём = new ProjectManager(
    "Артём",
    "Проектный менеджер",
    150000,
    "Отдел мобильной разработки",
    "Новое мобильное приложение"
);

Console.WriteLine();
Console.WriteLine("========== ИНФОРМАЦИЯ О СОТРУДНИКАХ ==========");

Console.WriteLine();
Console.WriteLine("--- Информация о простом сотруднике ---");
сотрудникИван.PrintInfo();
Console.WriteLine("---------------------------------------");

Console.WriteLine();
Console.WriteLine("--- Информация о менеджере ---");
менеджерЕгор.PrintInfo();
Console.WriteLine("---------------------------------------");

Console.WriteLine();
Console.WriteLine("--- Информация о проектном менеджере ---");
проектныйМенеджерАртём.PrintInfo();
Console.WriteLine("---------------------------------------");

Console.WriteLine();
Console.WriteLine("========== ТЕСТИРОВАНИЕ ИНТЕРФЕЙСОВ ==========");
Console.WriteLine();

// Добавляем членов команды
проектныйМенеджерАртём.AddTeamMember("Ирина (iOS-разработчик)");
проектныйМенеджерАртём.AddTeamMember("Сергей (Android-разработчик)");
проектныйМенеджерАртём.AddTeamMember("Максим (Тестировщик)");

// Получаем и выводим список команды (метод из ITeamLeader)
Console.WriteLine();
Console.WriteLine($"--- Список команды проекта «{проектныйМенеджерАртём.ProjectName}» ---");
List<string> members = проектныйМенеджерАртём.GetTeamMembers();
foreach (string member in members)
{
    Console.WriteLine($"{member}");
}

// Генерируем и выводим отчет (метод из IReportable)
Console.WriteLine();
Console.WriteLine("--- Генерация отчета ---");
string report = проектныйМенеджерАртём.GenerateReport();
Console.WriteLine(report);

Console.WriteLine();
Console.WriteLine("========================================");
Console.WriteLine("           ПРОГРАММА ЗАВЕРШЕНА");
Console.WriteLine("========================================");
```
