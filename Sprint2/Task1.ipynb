Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также сложное наследование.

```polyglot-notebook
public class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }

    public Animal(string name, int age)
    {
        Name = name;
        Age = age;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Кличка: {Name}, Возраст: {Age}");
    }
}

public class Dog : Animal
{
    public string Breed { get; set; }

    public Dog(string name, int age, string breed) : base(name, age)
    {
        Breed = breed;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
    }
}

public class Cat : Animal
{
    public string Color { get; set; }

    public Cat(string name, int age, string color) : base(name, age)
    {
        Color = color;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Окрас: {Color}");
    }
}

public class CharacterCat : Cat
{
    public string CharacterType { get; set; }

    public CharacterCat(string name, int age, string color, string characterType) 
        : base(name, age, color)
    {
        CharacterType = characterType;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Характер: {CharacterType}");
    }
}

// Проверяем работу
Dog simpleDog = new Dog("Вольт", 2, "Хаски");
simpleDog.DisplayInfo();

Console.WriteLine();

CharacterCat charCat = new CharacterCat("Мася", 3, "Рыжий", "Ласковый");
charCat.DisplayInfo();

Console.WriteLine();

Cat myCat = new Cat("Мурзик", 2, "Серый");
myCat.DisplayInfo();
```
