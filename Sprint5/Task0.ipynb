Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте при помощи List и Dictionary<TKey,TValue>.

```polyglot-notebook
public class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }

    public Animal(string name, int age)
    {
        Name = name;
        Age = age;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age}");
    }
}

public class Dog : Animal
{
    public string Breed { get; set; }

    public Dog(string name, int age, string breed)
        : base(name, age)
    {
        Breed = breed;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
    }
}

public class Cat : Animal
{
    public bool IsIndoor { get; set; }

    public Cat(string name, int age, bool isIndoor)
        : base(name, age)
    {
        IsIndoor = isIndoor;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Домашняя: {IsIndoor}");
    }
}

public class Bird : Animal
{
    public bool CanFly { get; set; }

    public Bird(string name, int age, bool canFly)
        : base(name, age)
    {
        CanFly = canFly;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Умеет летать: {CanFly}");
    }
}

// --- Работа с List<T> ---
Console.WriteLine("Список животных (List)");
List<Animal> animalList = new List<Animal>();

animalList.Add(new Dog("Вольт", 3, "Хаски"));
animalList.Add(new Cat("Мася", 4, true));
animalList.Add(new Bird("Пог", 2, true));

foreach (var animal in animalList)
{
    animal.DisplayInfo();
    Console.WriteLine();
}

// --- Работа с Dictionary<TKey, TValue> ---
Console.WriteLine("Словарь животных по номеру паспорта (Dictionary)");
Dictionary<string, Animal> animalDictionary = new Dictionary<string, Animal>();

animalDictionary.Add("DOG001", new Dog("Вольт", 3, "Хаски"));
animalDictionary.Add("CAT002", new Cat("Мася", 4, true));
animalDictionary.Add("BRD003", new Bird("Пог", 2, true));

if (animalDictionary.TryGetValue("CAT002", out Animal foundAnimal))
{
    Console.WriteLine("Найденное животное:");
    foundAnimal.DisplayInfo();
}
else
{
    Console.WriteLine("Животное не найдено");
}
```
