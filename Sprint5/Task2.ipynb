Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) с применением делегатов, лямбда-выражений и событий.

```polyglot-notebook
// Объявляем делегат для вывода информации
public delegate void DisplayAnimalDelegate();

public class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }

    // Делегат и событие для уведомления об изменениях
    public delegate void AnimalHandler(string message);
    public event AnimalHandler AnimalUpdated;

    public Animal(string name, int age)
    {
        Name = name;
        Age = age;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age}");
    }

    public void UpdateName(string newName)
    {
        Name = newName;
        // Вызываем событие после смены имени
        AnimalUpdated?.Invoke($"Имя животного изменено на: {newName}");
    }
}

public class Dog : Animal
{
    public string Breed { get; set; }

    public Dog(string name, int age, string breed)
        : base(name, age)
    {
        Breed = breed;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
    }
}

public class Cat : Animal
{
    public bool IsIndoor { get; set; }

    public Cat(string name, int age, bool isIndoor)
        : base(name, age)
    {
        IsIndoor = isIndoor;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Домашняя: {IsIndoor}");
    }
}

public class Bird : Animal
{
    public bool CanFly { get; set; }

    public Bird(string name, int age, bool canFly)
        : base(name, age)
    {
        CanFly = canFly;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Умеет летать: {CanFly}");
    }
}

// --- 1. Использование делегатов ---
Console.WriteLine("= Делегаты =");
Dog dog = new Dog("Вольт", 3, "Хаски");
Cat cat = new Cat("Мася", 4, true);

DisplayAnimalDelegate displayDog = dog.DisplayInfo;
DisplayAnimalDelegate displayCat = cat.DisplayInfo;

displayDog();
displayCat();

// --- 2. Использование лямбда-выражений ---
Console.WriteLine("\n= Лямбда-выражения =");
List<Animal> animals = new List<Animal>
{
    new Dog("Вольт", 3, "Хаски"),
    new Cat("Мася", 4, true),
    new Dog("Ампер", 1, "Дворняга"),
    new Bird("Пог", 2, true)
};

// Фильтруем только собак старше 2 лет с помощью лямбда-выражения
var adultDogs = animals.OfType<Dog>().Where(d => d.Age > 2);

foreach (var d in adultDogs)
{
    d.DisplayInfo();
}

// --- 3. Использование событий ---
Console.WriteLine("\n= События =");
static void OnAnimalUpdated(string message)
{
    Console.WriteLine(message);
}

Animal someAnimal = new Animal("Старое Имя", 5);
// Подписываемся на событие
someAnimal.AnimalUpdated += OnAnimalUpdated;
// Изменяем имя (сработает событие)
someAnimal.UpdateName("Ом");
```
