Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также применить явную реализацию интерфейса.

```polyglot-notebook
public interface IAnimal
{
    void MakeSound();
    void Eat();
}

public abstract class Animal : IAnimal
{
    public string Name { get; set; }

    public abstract void MakeSound();
    public abstract void Eat();
}

public class Dog : Animal
{
    public override void MakeSound() => Console.WriteLine($"{Name} лает: Гав");
    public override void Eat() => Console.WriteLine($"{Name} ест корм для собак.");
}

public class Cat : Animal, IAnimal
{
    public bool IsIndoor { get; set; }

    void IAnimal.MakeSound() => Console.WriteLine($"{Name} мяукает: Мяу");
    void IAnimal.Eat() => Console.WriteLine($"{Name} ест корм для кошек.");

    public override void MakeSound() => Console.WriteLine($"{Name} молчит.");
    public override void Eat() => Console.WriteLine($"{Name} ест обычную еду.");
}

public class Bird : Animal
{
    public override void MakeSound() => Console.WriteLine($"{Name} поёт: Чирик");
    public override void Eat() => Console.WriteLine($"{Name} клюёт зёрна.");
}

Cat catDirect = new Cat { Name = "Мася" };
catDirect.MakeSound();

IAnimal catViaInterface = catDirect;
catViaInterface.MakeSound();
```
