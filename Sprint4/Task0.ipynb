Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также наследование с применением интерфейсов.

```polyglot-notebook
public interface IAnimal
{
    void MakeSound();
    void Eat();
    int GetAge();
}

public abstract class Animal : IAnimal
{
    public string Name { get; set; }
    public string Species { get; set; }
    protected int age = 0;

    public virtual void MakeSound() => Console.WriteLine($"{Name} издаёт звук.");
    public virtual void Eat() => Console.WriteLine($"{Name} ест.");
    public virtual int GetAge() => age;
}

public class Dog : Animal
{
    public string Breed { get; set; }

    public override void MakeSound()
    {
        base.MakeSound();
        Console.WriteLine("Гав");
    }

    public void Fetch() => Console.WriteLine($"{Name} приносит палку.");
}

public class Cat : Animal
{
    public bool IsIndoor { get; set; }

    public override void MakeSound()
    {
        base.MakeSound();
        Console.WriteLine("Мяу");
    }

    public void Scratch() => Console.WriteLine($"{Name} царапается.");
}

public class Bird : Animal
{
    public bool CanFly { get; set; }

    public override void MakeSound()
    {
        base.MakeSound();
        Console.WriteLine("Чирик");
    }

    public void Fly() => Console.WriteLine(CanFly ? $"{Name} летает." : $"{Name} не летает.");
}

IAnimal myDog = new Dog { Name = "Вольт", Breed = "Хаски" };
myDog.MakeSound();
((Dog)myDog).Fetch();

IAnimal myBird = new Bird { Name = "Пог", CanFly = true };
myBird.MakeSound();
((Bird)myBird).Fly();
```
