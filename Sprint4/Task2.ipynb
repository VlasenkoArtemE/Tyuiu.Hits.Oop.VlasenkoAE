Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также реализовать управление зависимостями.

```polyglot-notebook
public interface IAnimal
{
    void MakeSound();
    void Eat();
    int GetLegsCount();
}

public abstract class Animal : IAnimal
{
    public string Name { get; set; }

    public abstract void MakeSound();
    public abstract void Eat();
    public abstract int GetLegsCount();
}

public class Dog : Animal
{
    public override void MakeSound() => Console.WriteLine("Гав");
    public override void Eat() => Console.WriteLine($"{Name} ест мясо.");
    public override int GetLegsCount() => 4;
}

public class Cat : Animal
{
    public override void MakeSound() => Console.WriteLine("Мяу");
    public override void Eat() => Console.WriteLine($"{Name} ест рыбу.");
    public override int GetLegsCount() => 4;
}

public class Bird : Animal
{
    public override void MakeSound() => Console.WriteLine("Чирик");
    public override void Eat() => Console.WriteLine($"{Name} клюёт зёрна.");
    public override int GetLegsCount() => 2;
}

public class AnimalService
{
    private readonly IAnimal _animal;

    public AnimalService(IAnimal animal)
    {
        _animal = animal;
    }

    public void UseAnimal()
    {
        _animal.MakeSound();
        _animal.Eat();
        Console.WriteLine($"У этого животного {_animal.GetLegsCount()} ноги/лапы.");
    }
}

IAnimal myDog = new Dog { Name = "Вольт" };
AnimalService dogService = new AnimalService(myDog);
dogService.UseAnimal();

IAnimal myBird = new Bird { Name = "Пог" };
AnimalService birdService = new AnimalService(myBird);
birdService.UseAnimal();
```
