```polyglot-notebook
using System;
using System.Collections.Generic;

public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    // Метод цветения
    public virtual void Цветение()
    {
        Console.WriteLine($"{Name} начинает цвести.");
    }

    // Метод роста
    public void Растет(float рост)
    {
        Height += рост;

        Console.WriteLine(
            $"{Name} вырос на {рост} см. Новая высота: {Height} см."
        );
    }

    // Метод полива
    public virtual void Полив()
    {
        Console.WriteLine(
            $"{Name} получил воду и чувствует себя хорошо."
        );
    }

    // Вывод информации
    public virtual void ПоказатьИнформацию()
    {
        Console.WriteLine($"Название: {Name}");
        Console.WriteLine($"Тип: {Type}");
        Console.WriteLine($"Высота: {Height} см");
    }
}

public class Flower : Plant
{
    public string Color { get; set; }
    public int PetalCount { get; set; }

    public string Состояние { get; private set; }

    public Flower(
        string name,
        float height,
        string color,
        int petalCount)
        : base(name, "Цветок", height)
    {
        Color = color;
        PetalCount = petalCount;
        Состояние = "Цветёт";
    }

    // Переопределение метода цветения
    public override void Цветение()
    {
        Console.WriteLine(
            $"{Name} распускается и радует глаз!"
        );

        Console.WriteLine(
            $"Цвет: {Color}, лепестков: {PetalCount}."
        );
    }

    // Увядание
    public void Увядать()
    {
        Состояние = "Увядает";

        Console.WriteLine(
            $"{Name} начал увядать."
        );

        Console.WriteLine(
            $"Состояние: {Состояние}"
        );
    }

    // Переопределение полива
    public override void Полив()
    {
        if (Name == "Роза")
        {
            Console.WriteLine(
                "Роза напилась воды и раскрыла свои лепестки."
            );
        }
        else
        {
            Console.WriteLine(
                $"{Name} получил воду и продолжает расти."
            );
        }
    }

    // Информация о цветке
    public override void ПоказатьИнформацию()
    {
        Console.WriteLine($"Название: {Name}");
        Console.WriteLine($"Тип: {Type}");
        Console.WriteLine($"Цвет: {Color}");
        Console.WriteLine($"Лепестков: {PetalCount}");
        Console.WriteLine($"Высота: {Height} см");
        Console.WriteLine($"Состояние: {Состояние}");
    }
}

public class Tree : Plant
{
    public int Age { get; set; }
    public string LeafType { get; set; }

    public Tree(
        string name,
        float height,
        int age,
        string leafType)
        : base(name, "Дерево", height)
    {
        Age = age;
        LeafType = leafType;
    }

    // Переопределение метода цветения
    public override void Цветение()
    {
        Console.WriteLine(
            $"{Name} покрывается новыми листьями."
        );

        Console.WriteLine(
            $"Возраст: {Age} лет, тип листьев: {LeafType}."
        );
    }

    // Увеличение возраста дерева
    public void Вырастить(int годы)
    {
        Age += годы;

        Console.WriteLine(
            $"{Name} стал старше на {годы} года."
        );

        Console.WriteLine(
            $"Теперь ему {Age} лет."
        );
    }

    // Полив дерева
    public override void Полив()
    {
        Console.WriteLine(
            $"{Name} получил воду и продолжает укрепляться."
        );
    }

    // Информация о дереве
    public override void ПоказатьИнформацию()
    {
        Console.WriteLine($"Название: {Name}");
        Console.WriteLine($"Тип: {Type}");
        Console.WriteLine($"Высота: {Height} см");
        Console.WriteLine($"Возраст: {Age} лет");
        Console.WriteLine($"Тип листьев: {LeafType}");
    }
}

public class Garden
{
    public string Name { get; set; }

    private List<Plant> Plants { get; set; }

    public Garden(string name)
    {
        Name = name;
        Plants = new List<Plant>();
    }

    // Добавление растения
    public void ДобавитьРастение(Plant plant)
    {
        Plants.Add(plant);
    }

    // Показать все растения
    public void ПоказатьСад()
    {
        Console.WriteLine();
        Console.WriteLine("========================================");
        Console.WriteLine($"             {Name}");
        Console.WriteLine("========================================");

        foreach (Plant plant in Plants)
        {
            Console.WriteLine();
            Console.WriteLine("----------------------------------------");

            plant.ПоказатьИнформацию();

            Console.WriteLine("----------------------------------------");
        }
    }

    // Полиморфный вызов цветения
    public void ВсеЦветут()
    {
        Console.WriteLine();
        Console.WriteLine("========== ЦВЕТЕНИЕ ==========");

        foreach (Plant plant in Plants)
        {
            plant.Цветение();
        }
    }

    // Полив всех растений
    public void ПолитьВсе()
    {
        Console.WriteLine();
        Console.WriteLine("========== ПОЛИВ ==========");

        foreach (Plant plant in Plants)
        {
            plant.Полив();
        }
    }
}

Garden садик = new Garden("САД АРТЁМА");

Flower гортензия = new Flower(
    "Ромашка",
    55,
    "Белая",
    8
);

Flower роза = new Flower(
    "Роза",
    70,
    "Красная",
    24
);

// СОЗДАНИЕ ДЕРЕВЬЕВ

Tree ель = new Tree(
    "Ель",
    300,
    12,
    "Хвойные"
);

Tree береза = new Tree(
    "Берёза",
    220,
    9,
    "Лиственные"
);

// ДОБАВЛЕНИЕ РАСТЕНИЙ В САД

садик.ДобавитьРастение(гортензия);
садик.ДобавитьРастение(роза);
садик.ДобавитьРастение(ель);
садик.ДобавитьРастение(береза);

// НАЧАЛЬНОЕ СОСТОЯНИЕ

Console.WriteLine("НАЧАЛЬНОЕ СОСТОЯНИЕ САДА");

садик.ПоказатьСад();

// ЦВЕТЕНИЕ — ПОЛИМОРФИЗМ

садик.ВсеЦветут();

// РОСТ РАСТЕНИЙ

Console.WriteLine();
Console.WriteLine("========== РОСТ ==========");

гортензия.Растет(7);
роза.Растет(6);
ель.Растет(20);
береза.Растет(10);

// ВОЗРАСТ ДЕРЕВЬЕВ

Console.WriteLine();
Console.WriteLine("========== ВОЗРАСТ ДЕРЕВЬЕВ ==========");

ель.Вырастить(3);
береза.Вырастить(2);

// ПОЛИВ

садик.ПолитьВсе();

// УВЯДАНИЕ РОЗЫ

Console.WriteLine();
Console.WriteLine("========== УВЯДАНИЕ ==========");

роза.Увядать();

// ФИНАЛЬНОЕ СОСТОЯНИЕ САДА

Console.WriteLine();
Console.WriteLine("========== ИТОГ ==========");

садик.ПоказатьСад();

// КОНЕЦ ПРОГРАММЫ

Console.WriteLine();
Console.WriteLine("Программа завершена.");
```
