```polyglot-notebook
// Базовый абстрактный класс Component
public abstract class Component
{
    private string name;
    private decimal price;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }
        set
        {
            if (value >= 0)
                price = value;
        }
    }

    public Component(string name, decimal price)
    {
        this.name = name;
        this.price = price;
    }

    // Метод для вывода информации
    public virtual string GetInfo()
    {
        return $"Компонент: {name}, Цена: {price} руб.";
    }

    // Перегрузка метода ToString()
    public override string ToString()
    {
        return GetInfo();
    }
}

// Интерфейс для установки компонента
public interface IInstallable
{
    void Install();
}

public class CPU : Component, IInstallable
{
    public CPU(string name, decimal price) : base(name, price) { }

    public override string GetInfo() => $"Процессор (CPU): {Name}, Цена: {Price} руб.";

    public void Install() => Console.WriteLine($"Процессор «{Name}» установлен в материнскую плату.");

    public static bool operator ==(CPU a, CPU b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price;
    }
    public static bool operator !=(CPU a, CPU b) => !(a == b);
}

public class RAM : Component, IInstallable
{
    public RAM(string name, decimal price) : base(name, price) { }

    public override string GetInfo() => $"Оперативная память (RAM): {Name}, Цена: {Price} руб.";

    public void Install() => Console.WriteLine($"Память «{Name}» вставлена в слот.");

    public static bool operator ==(RAM a, RAM b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price;
    }
    public static bool operator !=(RAM a, RAM b) => !(a == b);
}

public class GPU : Component, IInstallable
{
    public GPU(string name, decimal price) : base(name, price) { }

    public override string GetInfo() => $"Видеокарта (GPU): {Name}, Цена: {Price} руб.";

    public void Install() => Console.WriteLine($"Видеокарта «{Name}» установлена в слот.");

    public static bool operator ==(GPU a, GPU b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price;
    }
    public static bool operator !=(GPU a, GPU b) => !(a == b);
}

public class HDD : Component, IInstallable
{
    public HDD(string name, decimal price) : base(name, price) { }

    public override string GetInfo() => $"Жесткий диск (HDD): {Name}, Цена: {Price} руб.";

    public void Install() => Console.WriteLine($"Жесткий диск «{Name}» подключен.");

    public static bool operator ==(HDD a, HDD b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price;
    }
    public static bool operator !=(HDD a, HDD b) => !(a == b);
}

public class Computer
{
    private string name;
    private List<Component> components;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public Computer(string name)
    {
        this.name = name;
        this.components = new List<Component>();
    }

    // Generic-метод добавления компонента
    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        components.Add(component);
        component.Install(); // Вызываем установку
    }

    // Generic-метод удаления компонента
    public void RemoveComponent<T>(T component) where T : Component
    {
        components.Remove(component);
        Console.WriteLine($"Компонент «{component.Name}» удален.");
    }

    // Подсчет общей стоимости всех компонентов
    public decimal GetTotalPrice()
    {
        decimal total = 0;
        foreach (var item in components)
        {
            total += item.Price;
        }
        return total;
    }

    // Вывод информации о сборке
    public void PrintSpecs()
    {
        Console.WriteLine($"\nСборка ПК: {name}");
        foreach (var item in components)
        {
            Console.WriteLine($"- {item.GetInfo()}");
        }
        Console.WriteLine($"Общая стоимость: {GetTotalPrice()} руб.");
    }
}

// Дополнительное задание: Фабрика с использованием Generic
public static class ComputerFactory
{
    public static Computer CreatePCWithComponent<T>(string pcName, T baseComponent) where T : Component, IInstallable
    {
        Computer pc = new Computer(pcName);
        pc.AddComponent(baseComponent);
        return pc;
    }
}

// Создание экземпляров классов
CPU процессор1 = new CPU("Intel Core i5", 15000);
CPU процессор2 = new CPU("Intel Core i7", 17000);
CPU процессор3 = new CPU("AMD Ryzen 7", 25000);

RAM память = new RAM("16GB DDR4", 4000);
GPU видеокарта = new GPU("RTX 3060", 32000);
HDD диск = new HDD("1TB Seagate", 3500);

Console.WriteLine("========== ИНФОРМАЦИЯ О КОМПОНЕНТАХ ==========\n");
Console.WriteLine(процессор1.GetInfo());
Console.WriteLine(память.GetInfo());
Console.WriteLine(видеокарта.GetInfo());
Console.WriteLine(диск.GetInfo());

Console.WriteLine("\n========== ТЕСТИРОВАНИЕ ОПЕРАТОРОВ СРАВНЕНИЯ ==========\n");
Console.WriteLine($"процессор1 == процессор2 (одинаковые): {процессор1 == процессор2}");
Console.WriteLine($"процессор1 == процессор3 (разные):     {процессор1 == процессор3}");
Console.WriteLine($"процессор1 != процессор3:              {процессор1 != процессор3}");

Console.WriteLine("\n========== СБОРКА КОМПЬЮТЕРА (GENERIC МЕТОДЫ) ==========\n");
Computer мойПК = new Computer("Игровой ПК");

// Добавление компонентов
мойПК.AddComponent(процессор1);
мойПК.AddComponent(память);
мойПК.AddComponent(видеокарта);
мойПК.AddComponent(диск);

// Вывод состава и цены
мойПК.PrintSpecs();

Console.WriteLine("\n========== УДАЛЕНИЕ КОМПОНЕНТА ==========\n");
мойПК.RemoveComponent(диск);
Console.WriteLine($"Новая общая стоимость: {мойПК.GetTotalPrice()} руб.");

Console.WriteLine("\n========== ДОПОЛНИТЕЛЬНОЕ ЗАДАНИЕ (ФАБРИКА) ==========\n");
// Создание через обобщенный метод фабрики
Computer офисныйПК = ComputerFactory.CreatePCWithComponent("Офисный ПК", new CPU("Intel Core i3", 8000));
офисныйПК.PrintSpecs();

Console.WriteLine("\n========================================");
Console.WriteLine("           ПРОГРАММА ЗАВЕРШЕНА");
Console.WriteLine("========================================");
```
