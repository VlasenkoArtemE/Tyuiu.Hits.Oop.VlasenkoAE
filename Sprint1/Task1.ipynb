Ниже в блоке по примеру создайте базовый класс Animal и производные классы (2-3 например Dog, Cat и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также статические и обычными атрибуты и метода класса.

```polyglot-notebook
using System;

// Базовый класс Animal
public class Animal
{
    // Приватные поля
    private string _name;
    private int _age;

    // Свойство Name с геттером и сеттером
    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    // Свойство Age с геттером и сеттером
    public int Age
    {
        get { return _age; }
        set
        {
            if (value >= 0)
                _age = value;
            else
                throw new ArgumentOutOfRangeException(
                    "Возраст не может быть отрицательным!");
        }
    }

    // Обычный атрибут
    public string Color { get; set; }

    // Статический атрибут
    public static int TotalAnimals { get; private set; } = 0;

    // Конструктор
    public Animal()
    {
        TotalAnimals++;
    }

    // Обычный метод
    public void Eat()
    {
        Console.WriteLine($"{Name} ест.");
    }

    // Обычный метод
    public void GetOlder()
    {
        Age++;
        Console.WriteLine($"{Name} стал старше. Возраст: {Age}");
    }

    // Статический метод
    public static void DisplayTotalAnimals()
    {
        Console.WriteLine($"Общее количество животных: {TotalAnimals}");
    }
}


// Производный класс Dog
public class Dog : Animal
{
    // Дополнительное свойство
    private string _breed;

    public string Breed
    {
        get { return _breed; }
        set { _breed = value; }
    }

    // Конструктор
    public Dog(string name, int age, string color, string breed)
    {
        Name = name;
        Age = age;
        Color = color;
        Breed = breed;
    }

    // Обычный метод
    public void Bark()
    {
        Console.WriteLine($"{Name} говорит: Гав-гав!");
    }
}


// Производный класс Cat
public class Cat : Animal
{
    // Дополнительное свойство
    public bool IsIndoor { get; set; }

    // Конструктор
    public Cat(string name, int age, string color, bool isIndoor)
    {
        Name = name;
        Age = age;
        Color = color;
        IsIndoor = isIndoor;
    }

    // Обычный метод
    public void Meow()
    {
        Console.WriteLine($"{Name} говорит: Мяу!");
    }
}


// Производный класс Bird
public class Bird : Animal
{
    // Дополнительное свойство
    public double Wingspan { get; set; }

    // Конструктор
    public Bird(string name, int age, string color, double wingspan)
    {
        Name = name;
        Age = age;
        Color = color;
        Wingspan = wingspan;
    }

    // Обычный метод
    public void Fly()
    {
        Console.WriteLine($"{Name} летит.");
    }
}


// Создание объектов
Dog dog = new Dog("Вольт", 1, "Белый", "Хаски");
Cat cat = new Cat("Мася", 2, "Рыжий", true);
Bird bird = new Bird("Пог", 3, "Красный", 0.3);


// Вывод информации о животных
Console.WriteLine("=== Собака ===");
Console.WriteLine($"Имя: {dog.Name}");
Console.WriteLine($"Возраст: {dog.Age}");
Console.WriteLine($"Цвет: {dog.Color}");
Console.WriteLine($"Порода: {dog.Breed}");
dog.Eat();
dog.Bark();

Console.WriteLine();

Console.WriteLine("=== Кошка ===");
Console.WriteLine($"Имя: {cat.Name}");
Console.WriteLine($"Возраст: {cat.Age}");
Console.WriteLine($"Цвет: {cat.Color}");
Console.WriteLine($"Домашняя: {cat.IsIndoor}");
cat.Eat();
cat.Meow();

Console.WriteLine();

Console.WriteLine("=== Птица ===");
Console.WriteLine($"Имя: {bird.Name}");
Console.WriteLine($"Возраст: {bird.Age}");
Console.WriteLine($"Цвет: {bird.Color}");
Console.WriteLine($"Размах крыльев: {bird.Wingspan} м");
bird.Eat();
bird.Fly();

Console.WriteLine();

// Изменение свойств через сеттеры
dog.Age = 2;
dog.Color = "Серый";

Console.WriteLine($"{dog.Name}: новый возраст — {dog.Age}, новый цвет — {dog.Color}");


// Вызов обычного метода
cat.GetOlder();

// Вызов статического метода
Animal.DisplayTotalAnimals();
```
